```python
from pyspark.sql import SparkSession
 
spark = SparkSession.builder \
    .appName("PySparkPractice") \
    .getOrCreate()
 
data = [

    (101, "John Smith", "IT", "5000", "2026-01-15", "ACTIVE", "New York"),

    (102, "Mary Jones", "HR", "4500", "02/20/2026", "active", "Chicago"),

    (103, "David Lee", "IT", "6500", "2026-03-10", "INACTIVE", "Dallas"),

    (104, "Sara Khan", "Finance", None, "04/05/2026", "ACTIVE", "New York"),

    (105, "Mike Brown", "Finance", "7200", "2026-05-18", None, "Chicago"),

    (106, "Emma Davis", "HR", "4800", "06/25/2026", "ACTIVE", None),

    (107, "Chris Wilson", "IT", "8000", "2026-07-01", "active", "Dallas")

]
 
columns = [

    "employee_id",

    "employee_name",

    "department",

    "salary",

    "join_date",

    "status",

    "city"

]
 
df = spark.createDataFrame(data, columns)
 
df.show()

df.printSchema()
 
```

```python
import pyspark.sql.functions as F

df2 = df.withColumn("join_date", F.coalesce(F.try_to_date("join_date", "yyyy-MM-dd"), F.try_to_date("join_date", "MM/dd/yyyy")))
```

```python
df2.show()
```

```python
"""

"""
from pyspark.sql.functions import when
 

# when(col("salary") >= 7000, "HIGH")
# .when(col("salary") >= 5000, "MEDIUM")


from pyspark.sql.functions import (
    col,
    year,
    month,
    dayofmonth,
    to_date,
    when,
    upper
)

# Extract Year, Month and Day


df2.withColumn("join_year", year("join_date")) \
       .withColumn("join_month", month("join_date")) \
       .withColumn("join_day", dayofmonth("join_date"))


"""
Find employees who joined after March 2026.

"""


df2.filter(col("join_date") > "2026-03-31").show()


"""
Normalize status.

"""
df2 = df2.withColumn("status", upper(col("status")))


"""
Convert employee names to uppercase:

"""
df2 = df2.withColumn("employee_name", upper(col("employee_name")))


# Find rows containing null salary:

df2.filter(col("salary").isNull()).show()

# Find non-null salary:

df2.filter(col("salary").isNotNull()).show()


# Fill missing salary:

df2 = df2.fillna({
    "salary": 5000
})


# Fill multiple columns:

df2 = df2.fillna({
    "salary": 5000,
    "status": "UNKNOWN",
    "department": "NA"
})


# Rename Columns
df2 = df2.withColumnRenamed("employee_name", "emp_name")

# Delete / Drop Columns
df2 = df2.drop(
    "join_year",
    "join_month",
    "join_day"
)

# Create a salary category.
df2 = df2.withColumn(
    "salary_category",
    when(col("salary") >= 7000, "HIGH")
    .when(col("salary") >= 5000, "MEDIUM")
    .otherwise("LOW")
)
```

```python
df2.orderBy(col("salary").desc()).show()
```

```python
df2 = df2.withColumn("salary", col("salary").cast("int"))
```

```python
df2.groupBy("department").agg(F.sum("salary").alias("total_salary")).show()
```

```python
df2.agg(
    F.count("*").alias("employee_count"),
    F.avg("salary").alias("average_salary"),
    F.max("salary").alias("maximum_salary"),
    F.min("salary").alias("minimum_salary"),
    F.sum("salary").alias("total_salary")
).show()
```
